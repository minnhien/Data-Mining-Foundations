```python

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from scipy.stats import pearsonr
from sklearn.neighbors import KNeighborsRegressor
```

```python
train_data = pd.read_csv("train_3.csv")
test_data = pd.read_csv("test_3.csv")
train_reg = pd.read_csv("train_reg.csv")
test_reg = pd.read_csv("test_reg.csv")
```

```python

df_train = pd.concat([train_reg["averageRating"], train_data.drop(columns=['rating'])], axis=1)
df_test = pd.concat([test_reg["averageRating"], test_data.drop(columns=['rating'])], axis=1)
```

## Audience evaluation target feature: rating

```python
predictors = ['runtimeMinutes', 'awardWins','numVotes', 'titleType',  'numRegions','userReviewsTotal']
target = ["averageRating"]
```

# Simple Regression Univariate

### Coefficient: 
Represents the slope of the regression line for the predictor. A positive coefficient means the predictor is positively correlated with the target, while a negative coefficient indicates a negative correlation.
### Intercept: 
The value of the target variable (averageRating) when the predictor is zero. This serves as the starting point of the regression line.
### R² (Coefficient of Determination): 
Indicates the proportion of variance in the target variable explained by the predictor. Values closer to 1 suggest a stronger explanatory power. Negative or close-to-zero values indicate weak or no relationship.
### MSE (Mean Squared Error): 
Measures the average squared difference between the predicted and actual target values. Lower values indicate better model performance.
### MAE (Mean Absolute Error): 
Measures the average absolute difference between the predicted and actual target values. Like MSE, lower values indicate better performance.

### Linear Regression

```python
from sklearn.metrics import mean_absolute_error

# Initialize results dictionary
results = []

# Perform univariate regression and store detailed metrics
for predictor in predictors:
    # Prepare the data
    X_train = df_train[[predictor]]
    y_train = df_train[target]
    X_test = df_test[[predictor]]
    y_test = df_test[target]
    
    # Fit the model
    model = LinearRegression()
    model.fit(X_train, y_train)
    
    # Predict on test data
    y_pred = model.predict(X_test)
    
    # Evaluate the model
    mse = mean_squared_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    coef = model.coef_[0][0]
    intercept = model.intercept_[0]
    
    # Append the results
    results.append({
        'Predictor': predictor,
        'Coefficient': coef,
        'Intercept': intercept,
        'R²': r2,
        'MSE': mse,
        'MAE': mae
    })
results_df = pd.DataFrame(results)
```

```python
results_df
```

All predictors have very small coefficients, suggesting weak individual relationships with averageRating.
R² values close to 0 or negative indicate that none of the predictors explains a significant proportion of the variability in averageRating.
The high MSE and MAE values show that the models perform poorly when using each predictor individually.
Conclusion: No single predictor has a strong impact on averageRating.

### Ridge

```python
from sklearn.linear_model import Ridge

# Initialize results list for Ridge regression
ridge_results = []

# Perform univariate Ridge regression for each predictor
for predictor in predictors:
    # Prepare the data
    X_train = df_train[[predictor]]
    y_train = df_train[target]
    X_test = df_test[[predictor]]
    y_test = df_test[target]
    
    # Fit the Ridge regression model
    ridge_model = Ridge(alpha=1.0)
    ridge_model.fit(X_train, y_train)
    
    # Predict on test data
    y_pred = ridge_model.predict(X_test)
    
    # Evaluate the model
    mse = mean_squared_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    coef = ridge_model.coef_[0][0]
    intercept = ridge_model.intercept_[0]
    
    # Append the results
    ridge_results.append({
        'Predictor': predictor,
        'Coefficient': coef,
        'Intercept': intercept,
        'R²': r2,
        'MSE': mse,
        'MAE': mae
    })

# Convert Ridge results to DataFrame for display
ridge_results_df = pd.DataFrame(ridge_results)
```

```python
ridge_results_df
```

The results for Ridge Regression being nearly identical to those of Linear Regression

### Lasso

```python
from sklearn.linear_model import Lasso

# Initialize results list for Lasso regression
lasso_results = []

# Perform univariate Lasso regression for each predictor
for predictor in predictors:
    X_train = df_train[[predictor]]
    y_train = df_train[target]
    X_test = df_test[[predictor]]
    y_test = df_test[target]
    
    # Fit the Lasso regression model
    lasso_model = Lasso(alpha=0.1)
    lasso_model.fit(X_train, y_train)
    
    # Predict on test data
    y_pred = lasso_model.predict(X_test)
    
    # Evaluate the model
    mse = mean_squared_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    coef = lasso_model.coef_[0]
    intercept = lasso_model.intercept_
    
    # Append the results
    lasso_results.append({
        'Predictor': predictor,
        'Coefficient': coef,
        'Intercept': intercept,
        'R²': r2,
        'MSE': mse,
        'MAE': mae
    })

# Convert Lasso results to DataFrame for display
lasso_results_df = pd.DataFrame(lasso_results)
```

```python
lasso_results_df
```

#### Limited Predictive Power:
None of the predictors individually contribute significantly to predicting averageRating, as shown by their low R² values and unchanged MSE/MAE.
#### Feature Elimination: 
Lasso successfully identifies and eliminates weak predictors by shrinking their coefficients to zero, simplifying the model.
#### Similar Performance: 
The overall performance of Lasso is comparable to Linear and Ridge Regression, but it offers an additional benefit of identifying irrelevant features.

## Nonlinear regression

### Decision Tree Regressor

```python
from sklearn.tree import DecisionTreeRegressor

# Initialize results list for Decision Tree regression
tree_results = []

# Perform univariate Decision Tree regression for each predictor
for predictor in predictors:
    # Prepare the data
    X_train = df_train[[predictor]]
    y_train = df_train[target]
    X_test = df_test[[predictor]]
    y_test = df_test[target]
    
    # Fit the Decision Tree regression model
    tree_model = DecisionTreeRegressor(max_depth=9, min_samples_leaf=100,min_samples_split=5)
    tree_model.fit(X_train, y_train)
    
    # Predict on test data
    y_pred = tree_model.predict(X_test)
    
    # Evaluate the model
    r2 = r2_score(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    
    # Append the results
    tree_results.append({
        'Predictor': predictor,
        'R²': r2,
        'MSE': mse,
        'MAE': mae
    })

# Convert Decision Tree results to DataFrame for display
tree_results_df = pd.DataFrame(tree_results)
```

```python
tree_results_df
```

### KNN Regressor

```python
from sklearn.neighbors import KNeighborsRegressor

# Initialize results list for KNN regression
knn_results = []

# Perform univariate KNN regression for each predictor
for predictor in predictors:
    # Prepare the data
    X_train = df_train[[predictor]]
    y_train = df_train[target]
    X_test = df_test[[predictor]]
    y_test = df_test[target]
    
    # Fit the KNN regression model
    knn_model = KNeighborsRegressor(n_neighbors=5)
    knn_model.fit(X_train, y_train)
    
    # Predict on test data
    y_pred = knn_model.predict(X_test)
    
    # Evaluate the model
    r2 = r2_score(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    
    # Append the results
    knn_results.append({
        'Predictor': predictor,
        'R²': r2,
        'MSE': mse,
        'MAE': mae
    })

# Convert KNN results to DataFrame for display
knn_results_df = pd.DataFrame(knn_results)
```
